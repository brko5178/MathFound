```python
import PIL
from PIL import Image
import numpy as np
```

```python
image1 = Image.open('spongebob.png')
image1.size
```

```python
image2 = Image.open('hk2.png')
image2.size
```

```python
image1 = image1.resize((640,640), Image.Resampling.NEAREST)
```

```python
image1 = np.asarray(image1)
image2 = np.asarray(image2)
```

```python
def f(a, p1, p2):
    return a*p1+(1-a)*p2
```

```python
alphas = [0, 0.2, 0.4, 0.6, 0.8, 1]
```

```python
array1 = []
for a in alphas:
    fa = f(a, image1, image2)
    array1.append(fa)
```

```python
array1 = np.asarray(array1)
```

```python
array1[0].dtype
```

```python
uint8_array = array1.astype(np.uint8)
image1 = Image.fromarray(uint8_array[0])
image1.save('output1.png')

image2 = Image.fromarray(uint8_array[1])
image2.save('output2.png')

image2 = Image.fromarray(uint8_array[2])
image2.save('output3.png')

image2 = Image.fromarray(uint8_array[3])
image2.save('output4.png')

image2 = Image.fromarray(uint8_array[4])
image2.save('output5.png')

image2 = Image.fromarray(uint8_array[5])
image2.save('output6.png')
```
