```python
def to_binary(number):
    byte = []
    if number == 0:
        return '0'
    while number > 0:
        bit = number % 2
        byte.append(str(bit))
        number = number // 2
    byte.reverse()
    return "".join(byte)
        
```

```python
import math
```

```python
def function1(x, y):
    return math.tan(x)-y
```

```python
def bisection(a, b, e, p):
    rMin = a
    rmax = b
    fRmin = function1(rMin, p)
    rMid = 0.5*(rmax+rMin)
    while (rmax - rMin) > (2*e):
        fRmid = function1(rMid, p)
        if (fRmin*fRmid) < 0:
            rmax = rMid
        else:
            rMin = rMid
            fRmin = fRmid
        rMid = 0.5*(rmax+rMin)
    return rMid
```

```python
bisection(-((math.pi/2)), math.pi/2, 0.05, 0.5)
```
