```python
def to_binary(number): #decimal to binary converter
    byte = []
    if number == 0:
        return '0'
    while number > 0:
        bit = number % 2
        byte.append(str(bit))
        number = number // 2
    byte.reverse()
    return "".join(byte)
        
```

**Newton's method** 

```python
def functionP(x): #derivative of tanx = sec^2(x) sec = 1/cos
    return (1/math.cos(x))*(1/math.cos(x))
```

```python
def newt(x0, y, e, maxIters): #implementation of newton's method baased on provided pseudo code
    x = x0
    fx = e+1
    iter = 0
    while abs(fx) > e and iter < maxIters:
        iter += 1
        fx = function1(x, y)
        fpx = functionP(x)
        x = x - (fx/fpx)
    return x
```

```python
def original_func(x): #original function
    return (x*x)+((math.e**(2*x)))
```

```python
def minFunc(x): #function we want to minimize. Uses the distance calculation from the origin, squared
    return (x*x)+((x*x)+(math.e**(2*x)))**2
```

```python
def gold_section(a, b, e): #golden selection method based on provided pseudocode
    r = 0.5*(1+math.sqrt(5))
    xL = a
    xU = b
    d = (xU-xL)/r
    x1 = xU - d
    x2 = xL + d
    f1 = minFunc(x1)
    f2 = minFunc(x2)
    while (xU-xL)>(2*e):
        if f1<f2:
            xU=x2
            x2=x1
            f2=f1
            d = (xU-xL)/r
            x1=xU-d
            f1 = minFunc(x1)
        else:
            xL=x1
            x1=x2
            f1=f2
            d=(xU-xL)/r
            x2=xL+d
            f2 = minFunc(x2)
    return 0.5*(xU+xL)
```

```python
def newt2(x0, e, maxIters): #same as above using new functions for fx and fpx
    x = x0
    fx = e+1
    iter = 0
    while abs(fx) > e and iter < maxIters:
        iter += 1
        fx = minFunc(x)
        fpx = minFuncP(x)
        x = x - (fx/fpx)
    return x
```
